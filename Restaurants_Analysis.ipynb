## City Analysis

```python
# Top city with the highest number of restaurants

city_counts = df['City'].value_counts()
top_city = city_counts.idxmax()
top_city_count = city_counts.max()
```

```python
print(top_city, top_city_count)
```

```python
# Average rating restaurants in each city

city_avg_rating = df.groupby('City')['Aggregate_rating'].mean().round(2)

print(city_avg_rating)
```

```python
# Highest average rating

top_city = city_avg_rating.idxmax()
top_rating = city_avg_rating.max()

print(top_city, top_rating)
```

## Online Delivery

```python
# percentage of restaurants that offer online delivery

online_delivery_count = df['Has_Online_delivery'].value_counts()['Yes']

online_delivery_percentage = (online_delivery_count / total_restaurants) * 100

print(round(online_delivery_percentage, 2))
```

```python
# average ratings of restaurants with and without online delivery

delivery_rating = df.groupby('Has_Online_delivery')['Aggregate_rating'].mean().round(2)

print(delivery_rating)
```

## Restaurant Ratings

```python
# Analyze the distribution of aggregate ratings 
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.hist(df['Aggregate_rating'], bins=10, color='green')
plt.axvspan(3, 4, color='yellow', alpha=0.3)

plt.title('Distribution of Aggregate Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Restaurants')

plt.show()
```

```python
# Add column to group rating

df['Rating_range'] = pd.cut(df['Aggregate_rating'], bins=[0,2,3,4,5])

df['Rating_range'].value_counts()
```

## Cuisine Combination

```python
## Most common combinations of cuisines 

df_combination = df[df["Cuisines"].str.contains(',')]

df_combination["Cuisines"].value_counts().head(5)
```

```python
cuisine_stats = df.groupby('Cuisines').agg({
    'Aggregate_rating': 'mean',
    'Restaurant_ID': 'count'
}).rename(columns={'Restaurant_ID' : 'Restaurant_count'})
cuisine_stats = cuisine_stats[cuisine_stats['Restaurant_count'] > 50] # for enough sample

cuisine_stats.sort_values(by='Aggregate_rating', ascending=False).head(10)
```

```python
import folium

# Create map centered around average location
map_center = [df['Latitude'].mean(), df['Longitude'].mean()]

restaurant_map = folium.Map(location=map_center, zoom_start=5)

for index, row in df.head(100).iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=row['Restaurant_Name']
    ).add_to(restaurant_map)

restaurant_map
```

## Restaurant Chains

```python
# Restaurant Chains
chain_counts = df['Restaurant_Name'].value_counts()

chain_counts.head(10)
```

```python
chain_analysis = df.groupby('Restaurant_Name').agg({
    'Aggregate_rating': 'mean',
    'Restaurant_ID': 'count'
}).rename(columns={'Restaurant_ID': 'Restaurant_count'}).round(2)

chain_analysis = chain_analysis[chain_analysis['Restaurant_count'] > 1]

chain_analysis = chain_analysis[chain_analysis['Restaurant_count'] > 10]
```

```python
# Most popular chains
chain_analysis.sort_values(by='Restaurant_count', ascending=False).head(10)
```

```python
# Highest rated chains
chain_analysis.sort_values(by='Aggregate_rating', ascending=False).head(10)
```
